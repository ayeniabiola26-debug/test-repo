```python
#union()
set1={1, 2, 3,}
set2 = {4, 5, 6}
result = set2.union(set1)
print(result)
```

```python
#union for another method
result= set1 | set2
print(result)
```

```python
#intersection method
set1= {1,2,3,4}
set2= {4,5,6,7}
result = set2.intersection(set1)
print(result)
```

```python
#intersection alternative
result=set1 & set2
print(result)
```

```python
#difference
set1= {1,2,3,4}
set2= {3,4,5,6,}
result=set1.difference(set2)
print(result)
```

```python
#difference alternative
result= set1-set2
print(result)
```

```python
#symmetric difference
```

```python
set1= {1,2,3,4}
set2= {3,4,5,6,}
result=set2.symmetric_difference(set1)
print(result)
```

```python
#alternative to symmetric difference
result= set1 ^ set2
print(result)
```

```python
#set comparism
#issubset()
set1={1,2}
set2={1,2,3,4}
print(set1.issubset(set2))
```

```python
#alternative for issubset
print(set1 <= set2)
```

```python
#issuperset()
set1={1,2,3,4}
set2={1,2}
print(set1.issuperset(set2))
```

```python
#alternative for issuperset
print(set1 >= set2)
```

# RANDOM MODULE

```python
import random
```

```python
#Generating random numbers (random.random()) generayes a random decimal between 0.0 and 1.0
random_num= random.random()
print(random_num)
```

```python
#random.randint(a,b) in a range, generates a random integer between a and b (inclusive), it is either positive or negative
random_num= random.randint(1,10)
print(random_num)
```

```python
#random float in a range(random.uniform(a,b)) generate a random-floating point number in the given range
random_num=random.uniform(1,5)
print(random_num)
```

```python
#Random choices from a list, select a random element(random.choice())
names=['Abiola','Carina','Halimah','Esther','Salamat','jemilah','Zainab','Lola','Adeola','Funmi','Rukayat']
random_name=random.choice(names)
print(random_name)
```

```python
#select multiple random elements(random.sample())
two_random_names = random.sample(names, 2)
print(two_random_names)
```

```python
#randomly shuffle a list, (random.shuffle()) elements in a list randomly.
names=['Abiola','Carina','Halimah','Esther','Salamat','jemilah','Zainab','Lola','Adeola','Funmi','Rukayat']
random.shuffle(names)
print(names)
```

```python
x= random.randint(1,20)
print(x)
print(x**2)
```

```python
#comparism (relational) operator
x=10
y=5
print(x==y)
print(x!=y)
print(x>y)
print(x<y)
print(x>=y)
print(x<=y)
```

```python
#logical operator
x=5
y=10
z=15
print(x < y or x > z)

print(x > 4 and y < z)
print(x > 5 and y > 10)
print(not(x > 5))
```
